```python
import numpy as np
from scipy.fft import fft, fftfreq
import matplotlib.pyplot as plt

def calculate_spectrum(signal, sampling_frequency, single_sided=True):
    """
    Calculates the amplitude spectrum of a signal.

    Args:
        signal (np.ndarray): The time-domain signal.
        sampling_frequency (float): The sampling frequency of the signal (Hz).
        single_sided (bool): If True, returns the single-sided spectrum (positive frequencies).
                             If False, returns the full spectrum (positive and negative frequencies).

    Returns:
        tuple: A tuple containing:
            - frequencies (np.ndarray): Array of frequencies (Hz).
            - amplitudes (np.ndarray): Array of corresponding amplitude magnitudes.
    """
    N = len(signal) # Number of sample points
    T = 1.0 / sampling_frequency # Sample spacing

    # Perform the FFT
    yf = fft(signal)

    if single_sided:
        xf = fftfreq(N, T)[:N//2] # Frequencies for the positive half of the spectrum
        # Calculate the single-sided amplitude spectrum
        # np.abs(yf[0:N//2]) gets the magnitude of the positive frequencies
        amplitudes = 1.0/N * np.abs(yf[0:N//2])
    else:
        xf = fftfreq(N, T) # All frequencies
        amplitudes = 1.0/N * np.abs(yf)

    return xf, amplitudes
```

```python
import numpy as np
import matplotlib.pyplot as plt

# Frequência de amostragem (Hz)
fs = 44100

# Duração do sinal (segundos)
duracao = 1

# Vetor de tempo (discretização do tempo contínuo)
t = np.linspace(0, duracao, int(fs * duracao), endpoint=False)

# Frequências do sinal a serem analisadas
frequencias = [500, 5000, 10000, 50000]

for f0 in frequencias:

    # Geração do sinal x(t) = cos(2πft)
    x = np.cos(2 * np.pi * f0 * t)

    # Cálculo do espectro do sinal
    f, X = calculate_spectrum(x, fs)

    # Plot do espectro
    plt.figure()
    plt.plot(f, X)
    plt.title(f"Espectro do sinal para f = {f0} Hz")
    plt.xlabel("Frequência (Hz)")
    plt.ylabel("Amplitude")
    plt.grid()
    plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.signal import chirp

# Frequência de amostragem (Hz)
fs = 44100

# Duração do sinal (segundos)
duracao = 1

# Vetor de tempo
t = np.linspace(0, duracao, int(fs * duracao), endpoint=False)

# Tipos de varredura de frequência
tipos = ['linear', 'quadratic', 'logarithmic']

for tipo in tipos:

    # Geração do sinal chirp:
    # f0 -> frequência inicial
    # f1 -> frequência final
    # t1 -> tempo final da varredura
    x = chirp(t, f0=500, f1=10000, t1=duracao, method=tipo)

    # Cálculo do espectro do sinal
    f, X = calculate_spectrum(x, fs)

    # Plot do espectro
    plt.figure()
    plt.plot(f, X)
    plt.title(f"Espectro do chirp ({tipo})")
    plt.xlabel("Frequência (Hz)")
    plt.ylabel("Amplitude")
    plt.grid()
    plt.show()
```

```python
def subamostrar(x, M):
    """
    Realiza subamostragem de um sinal

    x -> vetor do sinal original
    M -> fator de subamostragem

    retorna:
    y -> sinal subamostrado
    """

    # Seleciona uma amostra a cada M (descarta as demais)
    y = x[::M]

    return y

# Fatores de subamostragem
fatores = [2, 4, 8]

for M in fatores:

    # Aplica subamostragem
    y = subamostrar(x, M)

    # Nova frequência de amostragem
    fs_novo = int(fs / M)

    # Normalizar o sinal (evita problema de reprodução)
    y = y / np.max(np.abs(y))

    # Cálculo do espectro do sinal subamostrado
    f, Y = calculate_spectrum(y, fs_novo)

    # Plot
    plt.figure()
    plt.plot(f, Y)
    plt.title(f"Espectro após subamostragem (M = {M})")
    plt.xlabel("Frequência (Hz)")
    plt.ylabel("Amplitude")
    plt.grid()
    plt.show()

    # Reprodução do áudio
    display(Audio(y.astype(np.float32), rate=fs_novo))
```

```python
from scipy.signal import resample

for M in fatores:

    # Novo número de amostras após redução
    N = len(x) // M

    # Reamostragem do sinal (interpolação + filtragem interna)
    y = resample(x, N)

    # Nova frequência de amostragem
    fs_novo = fs / M

    # Cálculo do espectro
    f, Y = calculate_spectrum(y, fs_novo)

    # Plot
    plt.figure()
    plt.plot(f, Y)
    plt.title(f"Resample (M = {M})")
    plt.xlabel("Frequência (Hz)")
    plt.ylabel("Amplitude")
    plt.grid()
    plt.show()

    # Reprodução
    display(Audio(y, rate=int(fs_novo)))
```

```python
def sobreamostrar(x, L):
    """
    Realiza sobreamostragem de um sinal

    x -> sinal original
    L -> fator de aumento

    retorna:
    y -> sinal sobreamostrado
    """

    # Cria vetor maior preenchido com zeros
    y = np.zeros(len(x) * L)

    # Insere as amostras originais a cada L posições
    y[::L] = x

    return y

# Fatores de sobreamostragem
fatores_L = [2, 4, 8]

for L in fatores_L:

    # Aplicação da sobreamostragem
    y = sobreamostrar(x, L)

    # Nova frequência de amostragem
    fs_novo = fs * L

    # Espectro
    f, Y = calculate_spectrum(y, fs_novo)

    # Plot
    plt.figure()
    plt.plot(f, Y)
    plt.title(f"Sobreamostragem (L = {L})")
    plt.xlabel("Frequência (Hz)")
    plt.ylabel("Amplitude")
    plt.grid()
    plt.show()

    # Reprodução
    display(Audio(y, rate=int(fs_novo)))
```

```python
from scipy.signal import resample
from IPython.display import Audio
import matplotlib.pyplot as plt

fatores_L = [2, 4, 8]

for L in fatores_L:

    # SOBREAMOSTRAGEM MANUAL

    y_manual = sobreamostrar(x, L)
    fs_manual = int(fs * L)

    # Normalização
    y_manual = y_manual / np.max(np.abs(y_manual))

    # Espectro
    f_m, Y_m = calculate_spectrum(y_manual, fs_manual)

    # RESAMPLE

    N = len(x) * L
    y_resample = resample(x, N)
    fs_resample = int(fs * L)

    # Normalização
    y_resample = y_resample / np.max(np.abs(y_resample))

    # Espectro
    f_r, Y_r = calculate_spectrum(y_resample, fs_resample)

    # PLOT SOBREPOSTO

    plt.figure(figsize=(10, 5))

    # Resample (laranja)
    plt.plot(f_r, Y_r, label="Resample (interpolado)", color="orange")

    # Manual (azul)
    plt.plot(f_m, Y_m, label="Manual (inserção de zeros)", color="blue")

    plt.title(f"Comparação dos espectros (M = {L})")
    plt.xlabel("Frequência (Hz)")
    plt.ylabel("Amplitude")
    plt.legend()
    plt.grid()
    plt.xlim(0, fs_manual/2)  # limita ao espectro útil

    plt.show()

    # ÁUDIO

    print(f"\n--- M = {L} ---")

    print("Manual:")
    display(Audio(y_manual.astype(np.float32), rate=fs_manual))

    print("Resample:")
    display(Audio(y_resample.astype(np.float32), rate=fs_resample))
```

```python
# Sinal contínuo (aproximação)

fs = 10_000_000  # alta frequência para simular contínuo
t = np.linspace(0, 0.01, int(fs * 0.01), endpoint=False)

# x(t) = cos(2000πt) + sin(5000πt)
x = np.cos(2000*np.pi*t) + np.sin(5000*np.pi*t)

# Plot (zoom)
plt.plot(t[:1000], x[:1000])
plt.title("Sinal no tempo")
plt.xlabel("Tempo (s)")
plt.ylabel("Amplitude")
plt.grid()
plt.show()

# Espectro

f, X = calculate_spectrum(x, fs)

plt.figure()
plt.plot(f, X)
plt.title("Espectro do sinal contínuo")
plt.xlabel("Frequência (Hz)")
plt.ylabel("Amplitude")
plt.grid()
plt.show()

# Amostragem em Nyquist

# Frequências:
# cos(2000πt) → 1000 Hz
# sin(5000πt) → 2500 Hz → maior frequência

fs_nyq = 2 * 2500  # 5000 Hz

t_n = np.linspace(0, 0.01, int(fs_nyq * 0.01), endpoint=False)

x_n = np.cos(2000*np.pi*t_n) + np.sin(5000*np.pi*t_n)

# Espectro discreto

f_n, X_n = calculate_spectrum(x_n, fs_nyq)

plt.figure()
plt.plot(f_n, X_n)
plt.title("Espectro após amostragem (Nyquist)")
plt.xlabel("Frequência (Hz)")
plt.ylabel("Amplitude")
plt.grid()
plt.show()

# Reconstrução

from scipy.signal import resample

# Reconstruindo para alta taxa
x_rec = np.zeros_like(t)

T = 1 / fs_nyq
# Largura da janela (quantas amostras considerar)
L = 1

for n in range(len(x_n)):

    # Limitar com janela retangular
    if n - L < 0:
        n_min = 0
    else:
        n_min = n - L

    if n + L > len(x_n):
        n_max = len(x_n)
    else:
        n_max = n + L

    # Soma das sinc truncadas
    for k in range(n_min, n_max):
        x_rec += x_n[k] * np.sinc((t - k*T)/T)

plt.figure(figsize=(10,5))

# Original (alta resolução)
x_cont = np.cos(2000*np.pi*t) + np.sin(5000*np.pi*t)

plt.plot(t[:1000], x_cont[:1000], label="Original", linewidth=2)

# Reconstruído
plt.plot(t[:1000], x_rec[:1000], '--', label="Reconstruído (sinc truncada)")

plt.legend()
plt.title("Reconstrução com interpolação sinc")
plt.xlabel("Tempo (s)")
plt.ylabel("Amplitude")
plt.grid()

plt.show()
```

```python
from google.colab import files
import scipy.io.wavfile as wav
from scipy.signal import convolve, resample
import numpy as np
import matplotlib.pyplot as plt
from IPython.display import Audio

# Upload dos arquivos

uploaded = files.upload()

nomes = list(uploaded.keys())
print("Arquivos enviados:", nomes)

# Identificação automática
arquivo_handel = None
arquivo_taca = None
arquivo_h = None

for nome in nomes:
    nome_lower = nome.lower()

    if "handel" in nome_lower:
        arquivo_handel = nome
    elif "taca" in nome_lower or "taça" in nome_lower:
        arquivo_taca = nome
    elif "banheiro" in nome_lower:
        arquivo_h = nome

if arquivo_handel is None or arquivo_taca is None or arquivo_h is None:
    raise ValueError("Envie: handel.wav, taca.wav e h_banheiro.wav")

# Leitura dos sinais

fs_handel, handel = wav.read(arquivo_handel)
fs_taca, taca = wav.read(arquivo_taca)
fs_h, h = wav.read(arquivo_h)

# Converter para float
handel = handel.astype(float)
taca = taca.astype(float)
h = h.astype(float)

# Normalização
handel = handel / np.max(np.abs(handel))
taca = taca / np.max(np.abs(taca))
h = h / np.max(np.abs(h))

# Ajuste de frequência (base: handel)

fs_base = fs_h

# Ajusta handel
if fs_handel != fs_base:
    fator = fs_base / fs_handel
    handel = resample(handel, int(len(handel) * fator))

# Ajusta taça
if fs_taca != fs_base:
    fator = fs_base / fs_taca
    taca = resample(taca, int(len(taca) * fator))

# CONVOLUÇÕES

# 1) handel no banheiro
y_handel = convolve(handel, h, mode='full')
y_handel = y_handel / np.max(np.abs(y_handel))

# 2) taça no banheiro
y_taca = convolve(taca, h, mode='full')
y_taca = y_taca / np.max(np.abs(y_taca))

# ESPECTRO

f_handel, Y_handel = calculate_spectrum(y_handel, fs_base)
f_taca, Y_taca = calculate_spectrum(y_taca, fs_base)

# Plot comparativo
plt.figure(figsize=(10,5))

plt.plot(f_handel, Y_handel, label="handel no banheiro", color="blue")
plt.plot(f_taca, Y_taca, label="taça no banheiro", color="orange")

plt.title("Espectros após convolução")
plt.xlabel("Frequência (Hz)")
plt.ylabel("Amplitude")
plt.legend()
plt.grid()
plt.xlim(0, fs_base/2)

plt.show()
```
